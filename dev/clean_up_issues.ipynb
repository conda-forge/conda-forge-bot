```python
import os

import github
```

```python
gh = github.Github(auth=github.Auth.Token(os.environ["GH_TOKEN"]))
```

```python
repo = gh.get_repo("conda-forge/conda-forge-bot")
```

```python
import re

import tqdm

title_re = re.compile(r"\[[0-9]{4}-[0-9]{2}-[0-9]{2}\] failed job bot-")

num_done = 0
for issue in tqdm.tqdm(
    repo.get_issues(state="all", creator="regro-cf-autotick-bot"), ncols=80
):
    if title_re.match(issue.title):
        # tqdm.tqdm.write(f"deleting issue {issue.number}: {issue.title}")
        gh.requester.graphql_named_mutation(
            mutation_name="deleteIssue",
            mutation_input={"issueId": issue.node_id},
            output_schema="clientMutationId",
        )
        num_done += 1

    if num_done == 50:
        break
```
