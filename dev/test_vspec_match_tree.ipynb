```python
from conda.models.version import (
    OPERATOR_START,
    VersionOrder,
    treeify,
    version_relation_re,
)


def version_trees_are_equiv(t1, t2):
    if isinstance(t1, tuple) and isinstance(t2, tuple):
        if len(t1) != len(t2):
            return False
        else:
            return all(version_trees_are_equiv(tv1, tv2) for tv1, tv2 in zip(t1, t2))
    elif isinstance(t1, str) and isinstance(t2, str):
        if t1 == t2:
            return True
        elif t1[0] in OPERATOR_START and t2[0] in OPERATOR_START:
            m1 = version_relation_re.match(t1)
            m2 = version_relation_re.match(t2)

            if m1 is None or m2 is None:
                return False
            else:
                op1, v1str = m1.groups()
                op2, v2str = m2.groups()

                if op1 != op2:
                    return False
                else:
                    return VersionOrder(v1str) == VersionOrder(v2str)
        else:
            return False
    else:
        return False
```

```python
version_trees_are_equiv(treeify("==3|(>1,==2)"), treeify("==3|(>1,==2)"))
```

```python
VersionOrder("1.*") == VersionOrder("1")
```
